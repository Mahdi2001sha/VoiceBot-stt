```python
import os

def show_tree(directory, prefix="", is_last=True):
    items = os.listdir(directory)
    folders = []
    files = []
    
    for item in items:
        full_path = os.path.join(directory, item)
        if os.path.isdir(full_path):
            folders.append(item)
        else:
            files.append(item)
    
    folders.sort()
    files.sort()
    sorted_items = folders + files
    
    for i, item in enumerate(sorted_items):
        is_last_item = (i == len(sorted_items) - 1)
        
        if is_last:
            print(prefix + "└── " + item)
        else:
            print(prefix + "├── " + item)
        
        full_path = os.path.join(directory, item)
        if os.path.isdir(full_path):
            new_prefix = prefix + ("    " if is_last else "│   ")
            show_tree(full_path, new_prefix, is_last_item)

# Run it for current directory
print(os.getcwd())
show_tree(".")
```
